# 1.3 Model Selection

In this example, we use the posterior pdf to discriminate between a point source and a distributed source.

```python
#__ Import: Numerical __
import numpy as np

#__ Import: Topophone2 __
from topophone2.topophone import decompose
from topophone2.inversion import MarginalStats
from topophone2.utils.read import read_txt
```

```python
#__ Set up: Random seed __
np.random.seed(2026)

#__ Disable: Warnings __
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
```

### 1.3.1 Create alternative characteristic function

In Example 1.1, we considered a distributed source on the seafloor, which is distributed in azimuth relative to the hydrophone array. A point source would not be distributed in azimuth. It woud, however, be distributed in apparent slowness as it excites multiple modes of the water column. We therefore set up a forward model that follows a normal distribution in apparent slowness only, as opposed to the bivariate distribution in Example 1.1.

First, we create a ``dict``, which is keyed by the name of the forward modeller (``'radial2D'``) pointing towards a ``tuple`` of the number of parameters (``3`` - the 2D mean and 1D scale) and the characteristic function ``cf_radial2D``. 

```python
#__ Apparent velocity beamwidth only __
def cf_radial2D(
    t   : np.ndarray,
    m_s : np.ndarray
):
    r"""

    log(cf) = i (t.xi) - (1/2) sigma^2 (t.xi)^2 / (xi.xi)

    Input
    -----
    `t` : `numpy.ndarray`
        Characteristic function argument, omega * R
        (frequency, sensor pair, direction)

    `m_s` : `numpy.ndarray`
        Model with mean slowness (X,Y) and radial beamwidth parameters
        (model, parameter)

    Output
    ------
    `cf` : `numpy.ndarray`
        Characteristic function
        (model, frequency, sensor pair)

    `dcf` : `numpy.ndarray`
        Gradient of characteristic function
        (model, frequency, sensor pair, parameter)
    """

    #__ np.einsum subscripts:
    #__ [i] model
    #__ [j] frequency
    #__ [k] sensor pair
    #__ [l],[p] direction

    #__ Unpack model vector __
    xi    = m_s[:, :2]  # (i,l)
    sigma = m_s[:, 2]   # (i,)

    #__ Invariants __
    u  = np.einsum('jkl,il->ijk', t, xi, optimize=True)  # (i,j,k)  t.xi
    s  = np.einsum('il,il->i', xi, xi, optimize=True)    # (i,)     xi.xi
    s3 = s[:, None, None]                                # (i,1,1)
    g  = u / s3                                          # (i,j,k)
    v2 = (sigma**2)[:, None, None]                       # (i,1,1)

    #__ Calculate characteristic function __
    L  = 1j * u
    Q  = -0.5 * v2 * u * g
    cf = np.exp(L + Q)

    #__ Calculate gradients __
    #__ d/dxi  = i t - sigma^2 (u/s) [ t - (u/s) xi ]
    #__ d/dsig = -sigma u^2 / s
    t_perp = t[None, ...] - g[..., None] * xi[:, None, None, :]     # (i,j,k,l)
    dlog          = np.empty(cf.shape + (m_s.shape[1],), dtype=cf.dtype)
    dlog[...]     = 0.0
    dlog[..., :2] = 1j * t[None, ...] - (v2 * g)[..., None] * t_perp
    dlog[..., 2]  = -sigma[:, None, None] * u * g
    dcf           = cf[..., None] * dlog

    #__ Return output __
    return cf, dcf

#__ `custom_cf` input __
custom_cf = {'radial2D':(
    3,           # Number of model parameters
    cf_radial2D  # Characteristic function
)}
```

### 1.3.2 Calculate posterior pdf under alternative parameterisation

We use the same observations as Example 1.1.

```python
#__ Read: Signal + noise __
out = read_txt("../data/eg1/triad/triad_topophone2.txt")
freq_Hz   = out[0]  # Frequency vector          [Hz]        (frequency)
x         = out[1]  # Receiver position         [m]         (receiver, direction)
Gamma_obs = out[2]  # Sample-covariance matrix  [uPa2.Hz-1] (frequency, receiver, receiver)
N_obs     = out[3]  # Independent observations
freq      = 2 * np.pi * freq_Hz  # Angular frequency vector [rad.s-1] (frequency)

#__ Read: Noise only __
out = read_txt("../data/eg1/triad/triad_noise-only_topophone2.txt")
Gamma_obs_noise = out[2]  # Array-covariance matrix  [uPa2.Hz-1] (frequency, receiver, receiver)
N_obs_noise     = out[3]  # Independent observations

#__ Noise apparent slowness [s.m-1] __
s0_J = 6.683784304186701e-4
```

In this example, the parameterisation is not a ``'normal'`` and ``'noise2D'`` component, but a ``'radial2D'`` and ``'noise2D'`` component. This is reflected in the list of signal types (``sigs``).

```python
#__ Signals present __
sigs = [
    'radial2D',                 # Alternative parameterisation
    dict(
        sig       = 'noise2D',  # Cylindrically-harmonic noise field
        s         = s0_J,       # Local apparent slowness [s.m-1]
        max_order = 1           # Maximum cylindrical-harmonic order
    )
]

#__ Prior bounds __
lower = np.array([                  # Lower bound
    50.,  -6.6e-4, 1.2e-4, 1.0e-6,  #     'radial2D' parameters [energy, xi_x, xi_y, sigma_r]
    50., -0.15, -0.15                 #     'noise2D' parameters  [energy, a_1, b_1]
])
upper = np.array([                  # Upper bound
    100., -5.8e-4, 2.0e-4, 5.0e-5,
    100., 0.35,  0.35
])

#__ Number of flows __
N_flow = 8

#__ Inversion parameters __
learning_rate = 5e-4   # Learning rate for the optimiser
N_iterations  = 30000  # Number of iterations for optimiser
N_montecarlo  = 100    # Number of samples to estimate ELBO

#__ Number of samples to draw __
N_samples = 20000
```

Instead of instantiating all the inversion classes manually, as in Example 1.1, we use the ``topophone.decompose`` helper function to run full workflow.

The custom characterisation function is included in the forward modeller by specifying the dictionary in the ``custom_cf`` option. Note that there is no limit to the number of custom characteristic functions a user can input.

```python
#__ Perform inversion __
decompose(
    Gamma_obs       = Gamma_obs,
    N_obs           = N_obs,
    Gamma_obs_noise = Gamma_obs_noise,
    N_obs_noise     = N_obs_noise,
    x               = x,
    freq            = freq,
    sigs            = sigs,
    custom_cf       = custom_cf,
    prior_bounds    = (lower,upper),
    N_flow          = N_flow,
    learning_rate   = learning_rate,
    outpath         = "../models/eg1/mod",
    N_iterations    = N_iterations,
    N_montecarlo    = N_montecarlo,
    N_out           = 6,
    verbose         = True,
    N_samples       = N_samples,
    calc_likelihood = True
)

#__ Read results __
m0      = np.loadtxt(f"../models/eg1/mod/model-samples_iteration-{N_iterations:05d}.txt")
lnp_dm0 = np.loadtxt(f"../models/eg1/mod/log-likelihood_iteration-{N_iterations:05d}.txt")
```

### 1.3.3 Calculate Bayes' factor between null and alternative parameterisations

To compare the two hypotheses - distributed source (Example 1.1) and point source (exmple 1.3) - we calculate the Bayes factor, which is the ratio of their Bayesian evidence.

First, we load the inversion results from Example 1.1.

```python
#__ Read results __
result1 = np.load('../models/eg1/triad/inversion_results.npz')
m1      = result1['m']             # Model samples  (sample, parameter)
lnp_dm1 = result1['lnp_dm']        # Log-likelihood (sample,)
bounds1 = result1['prior_bounds']  # Prior bounds   (parameter,)
```

Second, we calculate the evidence using the ``calc_evidence`` method of the ``inversion.MarginalStats`` class.

```python
#__ Instantiate stats objects __
stats0 = MarginalStats(m=m0, lnp_dm=lnp_dm0, prior_bounds=(lower,upper))
stats1 = MarginalStats(m=m1, lnp_dm=lnp_dm1, prior_bounds=bounds1)

#__ Calculate evidence __
lnp_d0, dlnp_d0 = stats0.calc_evidence()
lnp_d1, dlnp_d1 = stats1.calc_evidence()
```

Third, we calculate the ratio of the evidence.

```python
#__ Calculate Bayes' factor __
K  = lnp_d1 - lnp_d0
dK = np.sqrt(np.mean(dlnp_d1)**2 + np.mean(dlnp_d0)**2)
print(f"The Bayes' factor is ({K:.1f} +/- {dK:.1f}) nats.")
```

This result can be compared to the thresholds given by Kass & Raftery [1995]:

| $K$ | $\log_{10} K$ [dits] | $\ln K$ [nats] | Strength of evidence |
|---|---|---|---|
| 1.0 - 3.2 | 0.0 - 0.5 | 0.0 - 1.2 | Not worth more than a bare mention |
| 3.2 - 10.0 | 0.5 - 1 | 1.2 - 2.3 | Substantial |
| 10.0 - 100.0 | 1.0 - 2.0 | 2.3 - 4.6 | Strong |
| > 100.0 | > 2.0 | > 4.6 | Decisive |
