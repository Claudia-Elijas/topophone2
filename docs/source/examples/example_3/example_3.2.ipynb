## Loading the file

````python
#__ Parse configuration file __
args_dict = parse_config("../data/config_demo/config.ini")
````

## Accessing entries of the file

The parser function returns a dictionary with keys that correspond to the arguments of the ``topophone2.topophone.decomposition`` function, which performs probabilistic wavefield decomposition from a single call. 

````python
#__ Print parameters of configuration file __
for key, value in args_dict.items():
    print(f"{key} ")
````

We can access a given key as shown below,

````python
#__ Print example of entry __
N_obs = args_dict['N_obs'] ; print(f"N_obs: {N_obs}")
````

## Feeding configuration entries into the ``decomposition`` function

The keys returned by ``parse_config`` are in the same order as the arguments of ``decomposition``, we can simply input them as 

```python 

    decompose = (**args_dict)
```
